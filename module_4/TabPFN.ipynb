```python
from tabpfn import TabPFNRegressor
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

df = pd.read_csv('data/Processed_Data.csv')
df
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['StudentID', 'Semester', 'MDM', 'Stat', 'Formalise', 'Cost', 'Final_Exam_Score']), df['Final_Exam_Score'], test_size=0.3, random_state=42)
# convert everything to float
X_train = X_train.astype(float)
X_test = X_test.astype(float)
y_train = y_train.astype(float)
y_test = y_test.astype(float)
```

```python
X_train.fillna(0, inplace=True)
X_test.fillna(0, inplace=True)
```

```python
#pca
from sklearn.decomposition import PCA
pca = PCA(n_components=30)
pca.fit(X_train)
X_train = pca.transform(X_train)
X_test = pca.transform(X_test)
```

```python
reg = TabPFNRegressor(model_path='tabpfn-v3-regressor-v3_default.ckpt')
reg.fit(X_train, y_train)  # downloads checkpoint on first use
predictions = reg.predict(X_test)

print("--- TabPFN Regressor ---")
print(f"MSE: {mean_squared_error(y_test, predictions):.4f}")
print(f"R2: {r2_score(y_test, predictions):.4f}")
```
